```python
df = df.drop_duplicates()
```

```python
customer_df = df.dropna(subset=['CustomerID'])
```

```python
customer_df.shape
```

```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
```

```python
df.head()
```

```python
df['Revenue'].sum()
```

```python
country_revenue = df.groupby('Country')['Revenue'].sum()
```

```python
country_revenue
```

```python
country_revenue.sort_values(ascending=False).head(10)
```

```python
top_customers = df.groupby('CustomerID')['Revenue'].sum().sort_values(ascending=False)
```

```python
top_customers.head(10)
```

```python
import matplotlib.pyplot as plt
```

```python
top_customers.head(10).plot(kind='bar')
plt.xlabel('Customer ID')
plt.ylabel('Revenue (£)')
plt.title('Top 10 Customers by Revenue')
plt.show()
```

```python
top_products = (
    df.groupby('Description')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

top_products.head(10)
```

```python
top_quantity = (
    df.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

top_quantity.head(10)
```
