```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
```

```python
country_revenue = df.groupby('Country')['Revenue'].sum()
```

```python
country_revenue.sort_values(ascending=False).head(10)
```

```python
top_customers = df.groupby('CustomerID')['Revenue'].sum().sort_values(ascending=False)
```

```python
top_customers.head(10)
```

```python
import matplotlib.pyplot as plt
```

```python
top_customers.head(10).plot(kind='bar')
plt.xlabel('Customer ID')
plt.ylabel('Revenue (£)')
plt.title('Top 10 Customers by Revenue')
plt.show()
```

```python
top_products = (
    df.groupby('Description')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

top_products.head(10)
```

```python
top_quantity = (
    df.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

top_quantity.head(10)
```

```python
country_orders = (
    df.groupby("Country")["InvoiceNo"]
    .nunique()
    .sort_values(ascending=False)
)

country_orders.head(10)
```

```python
#Which countries place the highest number of orders?
country_orders = (
    df.groupby("Country")["InvoiceNo"]
    .nunique()
    .sort_values(ascending=False)
)

country_orders.head(10)
```

```python
#Which month generates the highest revenue?
df['Month'] = df['InvoiceDate'].dt.month
```

```python
df[["InvoiceDate", "Month"]].head()
```

```python
monthly_revenue = (
    df.groupby("Month")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

monthly_revenue
```

```python
monthly_revenue.sort_index().plot(kind="bar", figsize=(8,5))

plt.title("Monthly Revenue")

plt.xlabel("Month")

plt.ylabel("Revenue (£)")

plt.show()
```

```python
df['Day'] = df['InvoiceDate'].dt.day_name()
```

```python
highest_revenue_day = (
    df.groupby('Day')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

highest_revenue_day
```

```python
df['Hour'] = df['InvoiceDate'].dt.hour
```

```python
highest_revenue_hour = (
    df.groupby('Hour')['Revenue']
    .sum()
    .sort_values(ascending=False)

)

highest_revenue_hour
```

```python
#Which products sold the highest number of units?
top_products_quantity = (
    df.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

top_products_quantity.head(10)
```

```python
#which products have the highest number of returns?
returns = (
    df[df["Quantity"] < 0]
    .groupby("Description")["Quantity"]
    .sum()
    .sort_values()
)

returns.head(10)
```

```python
df[df["Quantity"] < 0]["Description"].value_counts().head(20)
```

```python
#Which country has the highest average revenue?
average_revenue_country = (
    df.groupby("Country")["Revenue"]
    .mean()
    .sort_values(ascending=False)
)

average_revenue_country.head(10)
```
